## 1. Listen, Tupel, Dictionaries und Sets

Dieses Beispiel zeigt mehrere wichtige Python-Datenstrukturen. Eine Liste ist veränderbar und kann Elemente hinzufügen oder entfernen. Ein Tupel ist dagegen unveränderlich. Ein Dictionary speichert Werte über Schlüssel, ähnlich einer Tabelle. Ein Set enthält eindeutige Elemente und unterstützt Operationen wie Vereinigung, Schnittmenge und Differenz.

```python
my_list = ["1", "pollutant", "stoff", "substanz"]
print(my_list[0]) # 0 bedeutet Anfang von list
print(my_list[-1]) # -1 bedeutet Ende von list
my_list[1] = "Banana"
print(my_list)
my_list.append("Orange") # fügt ein neues Element am Ende der Liste hinzu
print(my_list)
my_list.remove("Banana") # entfernt ein Element aus der Liste
print(my_list)

my_tuple = ("1", "pollutant", "stoff",  True) #die Tuple ist unveränderlich
print(my_tuple)


my_dick = {"1": "pollutant", "2": "stoff", "3": "substanz"} #DICTIONARY ist als excel Tabelle
print(my_dick)
print(my_dick["1"]) # gibt den Wert von key 1 aus

my_dick["1"] += ", Schadstoff" # fügt den Wert von key 1 hinzu
print(my_dick)
people = [
    {"name": "Max", "age": 28, "city": "New York"},
    {"name": "Bob", "age": 30, "city": "New Jearsey"}
    ] # LISTE von DICTIONARY
print(people)

people.append({"name": "Alice", 
               "age": 25, 
               "city": "Los Angeles"}) # fügt ein neues Dictionary in die Liste hinzu
print(people)
del people[1] # entfernt das zweite Dictionary aus der Liste
print(people)
del people[0]["city"] # entfernt den Wert von key "age" aus dem ersten Dictionary
print(people)
#Set 
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
set1.add(6) # fügt ein Element zum Set hinzu
print(set1)
set1.remove(3) # entfernt ein Element aus dem Set
print(set1)
print(set1.union(set2)) # gibt die Vereinigung von set1 und set2 aus
print(set1.intersection(set2)) # gibt die Schnittmenge von set1 und set2 aus
print(set1.difference(set2)) # gibt die Differenz von set1 und set2 aus


```

## 2. Strings formatieren

Hier werden drei Möglichkeiten gezeigt, Variablen in Text einzufügen: die Methode `.format()`, das Verbinden von Strings mit `+` und sogenannte f-Strings. F-Strings sind besonders übersichtlich, weil Variablen und Berechnungen direkt in geschweiften Klammern stehen können.

```python
first_name = "Max"
last_name = "Mustermann"
age = 28

print("{} {} ist {} Jahren alt".format(first_name, last_name, age)) # gibt den Namen und das Alter aus
print(first_name + " " + last_name + " ist " + str(age) + " Jahren alt") # gibt den Namen und das Alter aus. Nicht wirklich gut, da man die Variablen in Strings nicht direkt einfügen kann. Man muss sie vorher in Strings umwandeln.

#Docstring
print(f"{first_name} {last_name} ist {age} Jahren alt") # gibt den Namen und das Alter aus. Besser, da man die Variablen direkt in Strings einfügen kann.
print((f"Quadrat von 5 ist {5**2}")) # gibt das Quadrat von 5 aus. Besser, da man die Variablen direkt in Strings einfügen kann.
```

## 3. Mehrzeilige formatierte Texte

Mit dreifachen Anführungszeichen wird ein mehrzeiliger f-String erstellt. Die Variablen für Name, Kurs und Stunden werden direkt in den Text eingesetzt. Beim Ausgeben bleiben die Zeilenumbrüche erhalten.

```python
name = "Anna"
course = "Python"
hours = 10
info = f"""Student {name}
course: {course} 
hourse: {hours}"""
print(info) # gibt den Namen, den Kurs und die Stunden aus. Besser, da man die Variablen direkt in Strings einfügen kann. Man kann auch mehrere Zeilen in einem String haben.
```

## 4. Einen einfachen Kassenbon erstellen

Das Programm speichert Artikel, Preis und Menge in Variablen. Ein mehrzeiliger f-String erstellt daraus einen übersichtlichen Kassenbon. Der Gesamtpreis wird mit `price * qty` berechnet und direkt im Text angezeigt.

```python
item = "Coffee"
price = 2.5
qty = 2

receipt = f"""
==========================
Item: {item}
Price: {price}
Quantity: {qty}
Total: {price * qty}
=========================="""
print(receipt) # gibt den Namen, den Preis, die Menge und den Gesamtpreis aus. Besser, da man die Variablen direkt in Strings einfügen kann. Man kann auch mehrere Zeilen in einem String haben.
```

## 5. Zeilenumbrüche mit `\n`

Die Escape-Sequenz `\n` erzeugt innerhalb eines Strings einen Zeilenumbruch. Dadurch wird der Text bei der Ausgabe auf drei Zeilen verteilt.

```python
text =  "Line one \nLine two \nLine three"
print(text) # gibt den Text mit Zeilenumbrüchen aus. \n ist ein Zeilenumbruch
```

## 6. Mehrzeilige Strings direkt schreiben

Dreifache Anführungszeichen ermöglichen es, einen String über mehrere Codezeilen zu schreiben. Die enthaltenen Zeilenumbrüche werden automatisch gespeichert und bei `print()` ausgegeben.

## 7. Text mit einem Wahrheitswert ausgeben

Neben Name und Alter wird hier auch der Wahrheitswert `True` für den Studentenstatus gespeichert. Ein f-String setzt alle Werte in einen vollständigen Satz ein.

```python
First_name = "Joe"
Last_name = "Doe"
Age = 33
Student_status = True

print(f"{First_name} {Last_name} ist {Age} jahren alt. Ist {First_name} ein Student? {Student_status}")

```

## 8. Benutzereingaben verarbeiten

Das Programm fragt Vorname, Nachname, Alter und Studentenstatus ab. `int()` wandelt die Alterseingabe in eine ganze Zahl um. Der Vergleich mit `"True"` wandelt die Texteingabe in einen booleschen Wert um.

```python
First_name = input("Geben Sie Ihren Vornamen ein: ")
Last_name = input("Geben Sie Ihren Nachnamen ein: ")
Age = int(input("Geben Sie Ihr Alter ein: "))
Student_status = input("Sind Sie ein Student? (True/False): ") == "True"

print(f"{First_name} {Last_name} ist {Age} jahren alt. Ist {First_name} ein Student? {Student_status}")
```

## 9. Kurzschreibweisen für Änderungen

Mit `+=` wird eine Zahl erhöht und mit `-=` verringert. `x += 1` entspricht dabei `x = x + 1`, während `y -= 2` dasselbe wie `y = y - 2` bedeutet.

```python
x = 5
x += 1
print(x) # gibt 6 aus, da x um 1 erhöht wurde

y = 10
y -= 2
print(y) # gibt 8 aus, da y um 2 verringert wurde
```

## 10. Eine Zahl einlesen und ausgeben

Das Programm liest die Anzahl der Studenten ein. Mit `int()` wird die Eingabe von Text in eine ganze Zahl umgewandelt. Danach wird die Zahl mithilfe eines f-Strings ausgegeben.

```python
number_student = int(input("Geben Sie die Anzahl der Studenten ein: "))
print(f"Die Anzahl der Studenten beträgt: {number_student}")
```

## 11. Prüfen, ob eine Zahl positiv, negativ oder null ist

Mit `if`, `elif` und `else` wird der Wert von `Zahl` geprüft. Je nachdem, ob die Zahl größer als null, kleiner als null oder gleich null ist, wird eine passende Meldung ausgegeben.

```python
Zahl = 5
if Zahl > 0:
    print("Die Zahl ist positiv")
elif Zahl < 0:
    print("Die Zahl ist negativ")
else:
    print("Die Zahl ist null")
```
