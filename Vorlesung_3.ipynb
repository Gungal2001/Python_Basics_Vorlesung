## 1. Prüfen, ob eine Zahl positiv, negativ oder null ist

Dieses Programm liest eine ganze Zahl ein und speichert sie in `zahl`. Mit `if`, `elif` und `else` wird geprüft, ob die Zahl größer als null, kleiner als null oder genau null ist. Anschließend gibt das Programm eine passende Meldung aus.

```python
zahl = int(input("Geben Sie eine Zahl ein: "))

if zahl > 0:
    print(f"{zahl} ist positiv")
elif zahl < 0:
    print(f"{zahl} ist negativ")
else:
    print(f"{zahl} ist null")
    
```

## 2. Prüfen, ob eine Zahl gerade oder ungerade ist

Die eingegebene Zahl wird mit dem Modulo-Operator `%` durch 2 geteilt. Ist der Rest `0`, ist die Zahl gerade. Andernfalls ist sie ungerade.

```python
zahl = int(input("Geben Sie eine Zahl ein: "))

if zahl % 2 == 0:
    print(f"{zahl} ist gerade")
else:
    print(f"{zahl} ist ungerade")
```

## 3. Zwei Zahlen vergleichen

Das Programm liest zwei ganze Zahlen ein und vergleicht sie. Es gibt aus, welche Zahl größer ist oder ob beide Zahlen gleich groß sind.

```python
a = int(input("Erste Zahl: "))
b = int(input("Zweite Zahl: ")) 

if a > b:
    print(f"{a} ist größer als {b}")
elif a < b:
    print(f"{a} ist kleiner als {b}")
else:
    print("Beide Zahlen sind gleich")
```

## 4. Einfache Anmeldung prüfen

Hier werden ein Benutzername und ein Passwort eingelesen. Mit `and` wird überprüft, ob beide Angaben gleichzeitig den erwarteten Werten entsprechen. Nur dann ist der Login erfolgreich.

```python
benutzer = input("Geben Sie Ihren Benutzernamen ein: ")
passwort = input("Geben Sie Ihr Passwort ein: ")

if benutzer == "admin" and passwort == "1234":
    print("Login erfolgreich")
else:
    print("Falsche Zugangsdaten")
    
```

## 5. Mit einer `for`-Schleife zählen

Die `for`-Schleife durchläuft die Werte von `range(1, 6)`. Dabei werden die Zahlen 1 bis 5 nacheinander in `i` gespeichert und ausgegeben. Die 6 gehört nicht mehr zum Bereich, weil das Ende bei `range` ausgeschlossen wird.

```python
for i in range(1, 6): #i ist Variable, die von 1 bis 5 geht. Ersten Wert ist 1, Letzten Wert ist 6, da 6 nicht mitgezählt wird.
    print(i)
```

## 6. Bis zu einer eingegebenen Zahl zählen

Das Programm liest eine Endzahl ein. Danach zählt die `for`-Schleife von 1 bis zu dieser Zahl. Durch `ende + 1` wird die eingegebene Endzahl in den Bereich eingeschlossen.

```python
ende = int(input("Geben Sie eine Zahl ein, bis zu der gezählt werden soll: "))
for i in range(1, ende + 1):
    print(i)
```

## 7. Zeichen einer Zeichenkette ausgeben

Die `for`-Schleife geht jedes einzelne Zeichen der Zeichenkette `"Python"` durch. Das aktuelle Zeichen wird in `char` gespeichert und jeweils in einer eigenen Zeile ausgegeben.

```python
for char in "Python":
    print(char) #print(char) gibt jeden Buchstaben in der Zeichenkette "Python" aus.
```

## 8. Countdown mit einer `for`-Schleife

Mit `range(10, 0, -1)` wird rückwärts von 10 bis 1 gezählt. Die Schrittweite `-1` verringert die Zahl bei jedem Durchlauf um eins. Nach der Schleife wird `Start!!!!` ausgegeben.

```python
for i in range(10,0,-1): #Zählt von 10 bis 1 runter. Letzter Wert ist 0, da 0 nicht mitgezählt wird.
    print(i)

print("Start!!!!")
```

## 9. Vokale in einem Text zählen

Der eingegebene Text wird mit `lower()` in Kleinbuchstaben umgewandelt. Die Schleife untersucht jedes Zeichen. Wenn das Zeichen ein Vokal aus `aeiou` ist, wird `anzahl` um eins erhöht. Am Ende wird die Anzahl der gefundenen Vokale ausgegeben.

```python
text = input("Text: ").lower() #lower() wandelt den Text in Kleinbuchstaben um.
anzahl = 0

for zeichnen in text:
    if zeichnen in "aeiou":
        anzahl += 1

print(f"Der Text enthält {anzahl} Vokale.")
```

## 10. Countdown mit einer `while`-Schleife

Die Variable `count` startet mit dem Wert 5. Solange `count > 0` gilt, wird der aktuelle Wert ausgegeben und anschließend mit `count -= 1` um eins verringert. Wenn `count` den Wert 0 erreicht, endet die Schleife und es wird `Fertig!` ausgegeben.

```python
count = 5
while count > 0:
    print(count)
    count -= 1

print("Fertig!")
```
